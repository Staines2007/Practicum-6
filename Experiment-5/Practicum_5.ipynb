```python
import requests
import pandas as pd
import matplotlib.pyplot as plt


# 1. Retrieve the latest exchange rates
url = "https://api.frankfurter.app/latest"


response = requests.get(url)
data = response.json()


print("Base Currency:", data["base"])
print("Date:", data["date"])




# 2. Extract currency codes and exchange rates
rates = data["rates"]


print("\nAvailable Currency Codes and Exchange Rates:")
for currency, rate in rates.items():
    print(currency, ":", rate)




# 3. Convert JSON response into a DataFrame
df = pd.DataFrame(
    list(rates.items()),
    columns=["Currency", "Exchange_Rate"]
)


print("\nDataFrame:")
print(df)




# 4. Identify the five highest exchange rate values
top_5 = df.nlargest(5, "Exchange_Rate")


print("\nTop 5 Highest Exchange Rates:")
print(top_5)




# 5. Compare USD, EUR, INR, GBP, and JPY
selected_currencies = ["USD", "EUR", "INR", "GBP", "JPY"]


comparison = df[df["Currency"].isin(selected_currencies)]


print("\nSelected Currency Comparison:")
print(comparison)




# 6. Save data as exchange_rates.csv
df.to_csv("exchange_rates.csv", index=False)


print("\nData saved as exchange_rates.csv")




# 7. Visualize exchange rates using a bar chart
plt.figure(figsize=(12, 6))


plt.bar(df["Currency"], df["Exchange_Rate"])


plt.xlabel("Currency")
plt.ylabel("Exchange Rate")
plt.title("International Currency Exchange Rates")


plt.xticks(rotation=90)
plt.tight_layout()


plt.show()




# 8. Display strongest and weakest currencies
strongest = df.loc[df["Exchange_Rate"].idxmax()]
weakest = df.loc[df["Exchange_Rate"].idxmin()]


print("\nStrongest Currency:")
print("Currency:", strongest["Currency"])
print("Exchange Rate:", strongest["Exchange_Rate"])


print("\nWeakest Currency:")
print("Currency:", weakest["Currency"])
print("Exchange Rate:", weakest["Exchange_Rate"])


print("URK25CS7102")
```
